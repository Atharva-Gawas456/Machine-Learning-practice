```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
device = (
    "cuda" if torch.cuda.is_available()
    else "mps" if torch.backends.mps.is_available()
    else "cpu"
)
```

```python
print(f"Using device: {device}")
```

```python
x = torch.rand(3, 3, device = device)
y = torch.ones(3, 3, device = device)
z = torch.matmul(x, y)
```

```python
w = torch.tensor([2.0], requires_grad = True, device = device)
b = torch.tensor([1.0], requires_grad = True, device = device)
inputs = torch.tensor([3.0], device = device)

loss = (w * inputs + b) - 0.0
loss.backward()

print(f"Gradient with respect to w: {w.grad}")
print(f"Gradient with respect to b: {b.grad}")
```

```python
class TabularMLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(TabularMLP, self).__init__()

        self.network = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden_dim, output_dim)
        )

    def forward(self, x):
        return self.network(x)

model = TabularMLP(input_dim = 10, hidden_dim = 64, output_dim = 2).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(), lr = 0.001, weight_decay = 0.01)
```
