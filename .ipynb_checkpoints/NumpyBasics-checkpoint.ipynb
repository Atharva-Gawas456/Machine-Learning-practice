## Expressing Conditional Logic as Array Operations

```python
xarr = np.array([1.1, 1.2, 1.3, 1.4, 1.5])
yarr = np.array([2.1, 2.2, 2.3, 2.4, 2.5])
cond = np.array([True, False, True, True, False])
```

```python
result = [(x if c else y)
         for x, y, c in zip(xarr, yarr, cond)]
result
```

# 4.7 Example: Random Walks

```python
import random
import matplotlib.pyplot as plt

position = 0
walk = [position]
nsteps = 1000

for _ in range(nsteps):
    step = 1 if random.randint(0, 1) else -1
    position += step
    walk.append(position)

plt.plot(walk)
```

```python
nsteps = 1000
rng = np.random.default_rng(seed=12345)

draws = rng.integers(0, 2, size = nsteps)

steps = np.where(draws == 0, 1, -1)

walk = steps.cumsum()

plt.plot(walk)
```

```python
walk.min()
```

```python
walk.max()
```

```python
(np.abs(walk) >= 10).argmax()
```

## Simulating Many Random Walks at Once

```python
nwalks = 1000
nsteps = 1000

draws = rng.integers(0, 2, size=(nwalks, nsteps))

steps = np.where(draws > 0, 1, -1)

walks = steps.cumsum(axis = 1)

walks
```

```python
walks.max()
```

```python
walks.min()
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

alpha = 2.5

print(f"a = {a}")
print(f"b = {b}")

print("Addition Operation of Vector: ", a+b)
print("Scalar Operation of Vector: ", alpha*a)

dot_product = np.dot(a, b)
print("Dot Product of Vector: ", dot_product)

norm_a = np.linalg.norm(a)
print("Norm of Vector a: ", norm_a)

norm_b = np.linalg.norm(b)
print("Norm of Vector b:", norm_b)

cos_sim = np.dot(a, b)/(norm_a*norm_b)
print("Cosine Similarity", cos_sim)

cross_prod = np.cross(a, b)
print("Cross Product of Vectr: ", cross_prod)
```

```python
import numpy as np
import matplotlib.pyplot as plt


def plt_transform(ax, T, title, color='royalblue'):

    # Original unit square
    square = np.array([
        [0, 0],
        [1, 0],
        [1, 1],
        [0, 1],
        [0, 0]
    ]).T

    # Apply transformation
    transformed = T @ square

    # Basis vectors after transformation
    e1 = T @ np.array([1, 0])
    e2 = T @ np.array([0, 1])

    # Plot original square
    ax.plot(*square, 'k--', alpha=0.4, lw=1.0, label='Original')
    ax.fill(*square, alpha=0.08, color='gray')

    # Plot transformed square
    ax.plot(*transformed, lw=2.5, color=color, label=title)
    ax.fill(*transformed, alpha=0.15, color=color)

    # Plot transformed basis vectors
    ax.annotate(
        '',
        xy=e1,
        xytext=[0, 0],
        arrowprops=dict(
            arrowstyle="->",
            color="tomato",
            lw=2.5
        )
    )

    ax.annotate(
        '',
        xy=e2,
        xytext=[0, 0],
        arrowprops=dict(
            arrowstyle="->",
            color="green",
            lw=2.5
        )
    )

    # Axes styling
    ax.axhline(0, color="gray", linestyle="--", lw=0.8)
    ax.axvline(0, color="black", linestyle="--", lw=0.8)

    ax.set_title(title, fontsize=13, fontweight="bold")

    ax.legend()
    ax.grid(True, alpha=0.3)

    ax.set_aspect('equal')

    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)

    ax.set_xlabel('X1')
    ax.set_ylabel('X2')


# Rotation angle
theta = np.pi / 4

# Transformation matrices
transforms = [

    (
        np.array([
            [2, 0],
            [0, 2]
        ]),
        "Scaling Matrix"
    ),

    (
        np.array([
            [np.cos(theta), -np.sin(theta)],
            [np.sin(theta),  np.cos(theta)]
        ]),
        "Rotation Matrix"
    ),

    (
        np.array([
            [1, 1],
            [0, 1]
        ]),
        "Shear Matrix"
    )
]

# Create subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Colors for each transformation
colors = ['royalblue', 'tomato', 'green']

# Plot all transformations
for ax, (T, title), colour in zip(axes.flatten(), transforms, colors):

    plt_transform(ax, T, title, colour)

# Main title
plt.suptitle(
    "Transformation Matrix",
    fontsize=13,
    fontweight="bold"
)

plt.tight_layout()

# Show plot
plt.show()
```
