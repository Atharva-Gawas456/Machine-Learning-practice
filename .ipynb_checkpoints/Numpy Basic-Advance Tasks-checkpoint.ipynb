```python
import numpy as np
```

### 1. Array creation
Create 1D and 2D arrays from lists. Also create arrays using arange, linspace, zeros, ones, and eye.

```python
list1 = [1, 2, 3, 4]

list2 = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

arr1 = np.array(list1)

arr2 = np.array(list2)

print(arr1)
print("="*20)

print(arr2)
print("="*20)

arr_arange = np.arange(5)
print(arr_arange)

print("="*20)

arr_linspace = np.linspace(4,5)
print(arr_linspace)

print("="*20)

arr0 = np.zeros((5,5))
print(arr0)

print("="*20)

arr_1 = np.ones((5,5))
print(arr_1)

print("="*20)

arr_eye = np.eye(5)
print(arr_eye)
```

```python
print(arr2[1][2])

print("="*20)

print(arr2[1])

print("="*20)

print(arr2[:, 0])

print("="*20)

subarr = arr2[2]

print(subarr)
```

Compute mean, median, standard deviation, min, max, and percentiles on a 1D array.

```python
print(np.mean(arr2))

print("="*20)

print(np.median(arr2))

print("="*20)

print(np.std(arr2))

print("="*20)

print(np.minimum(arr2, arr3))

print("="*20)

print(np.maximum(arr2, arr3))
```
