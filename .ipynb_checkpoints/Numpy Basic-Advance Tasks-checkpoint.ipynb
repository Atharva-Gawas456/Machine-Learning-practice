### 1. Array creation
Create 1D and 2D arrays from lists. Also create arrays using arange, linspace, zeros, ones, and eye.

```python
list1 = [1, 2, 3, 4]

list2 = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

arr1 = np.array(list1)

arr2 = np.array(list2)

print(arr1)
print("="*20)

print(arr2)
print("="*20)

arr_arange = np.arange(5)
print(arr_arange)

print("="*20)

arr_linspace = np.linspace(4,5)
print(arr_linspace)

print("="*20)

arr0 = np.zeros((5,5))
print(arr0)

print("="*20)

arr_1 = np.ones((5,5))
print(arr_1)

print("="*20)

arr_eye = np.eye(5)
print(arr_eye)
```

## 6. Array sorting
Sort a 1D array in ascending and descending order. Also find the indices that would sort it using argsort.

```python
arr = np.array([5, 2, 7, 4, 1, 6])

print(np.sort(arr))

print("="*20)

print(np.sort(arr)[::-1])

print("="*20)

idx = np.argsort(arr)
print("argsort: ", idx)
print("Sorted via argsort: ", arr[idx])

print("="*20)

top3_idx = np.argsort(arr)[-3:][::-1]
print(top3_idx)
print(arr[top3_idx])

print("="*20)

arr.sort()
print(arr)
```
