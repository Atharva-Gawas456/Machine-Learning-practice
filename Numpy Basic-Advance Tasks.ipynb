## 6. Array sorting
Sort a 1D array in ascending and descending order. Also find the indices that would sort it using argsort.

```python
arr = np.array([5, 2, 7, 4, 1, 6])

print(np.sort(arr))

print("="*20)

print(np.sort(arr)[::-1])

print("="*20)

idx = np.argsort(arr)
print("argsort: ", idx)
print("Sorted via argsort: ", arr[idx])

print("="*20)

top3_idx = np.argsort(arr)[-3:][::-1]
print(top3_idx)
print(arr[top3_idx])

print("="*20)

arr.sort()
print(arr)
```

## 1. Reshaping & flattening
Reshape a 1D array into 2D and 3D. Flatten and ravel it back. Understand the difference between the two.

```python
a = np.arange(12)

b = a.reshape(3, 4)
c = a.reshape(2, 2, 3)

d = a.flatten()
e = a.ravel()

print(a)
print("="*20)

print(b)
print("="*20)

print(c)
print("="*20)

print(d)
print("="*20)

print(e)
print("="*20)
```

## 2. Stacking & splitting
Stack two arrays using vstack, hstack, and concatenate. Split an array into equal parts using array_split.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(np.vstack([a, b]))
print(np.hstack([a, b]))

print(np.concatenate([a, b], axis = 1))
print(np.concatenate([a, b], axis = 0))

x = np.arange(12)

parts = np.array_split(x, 3)
print(parts)

print(np.array_split(x, [3, 8]))
```

## 3. Broadcasting
Add a 1D array to each row of a 2D array without loops. Normalize a matrix row-wise using broadcasting.

```python
M = np.ones([3, 4])
v = np.array([10, 20, 30, 40])

print(M + v)

B = np.random.randint(1, 10, (4, 3))
row_sum = B.sum(axis = 1, keepdims = True)

normalized = B / row_sum

print(normalized)
```

## 4. Boolean masking
Filter elements from an array using conditions. Apply compound conditions using & and | operators.

```python
a = np.array([2, 5, 3, 1, 7, 0, 6])

mask = a > 4

print(a[mask])
print(a[a > 5])

print(a[(a > 2) & (a < 6)])
print(a[(a < 3) | (a > 5)])
print(a[~(a > 5)])
```

## 5. Fancy indexing
Select specific rows and columns using integer arrays as indices. Modify selected elements in-place.

```python
a = np.array([10, 20, 30, 40, 50])

idx = [0, 2, 4]

print(a[idx])
print(a[[-1, 1, 3]])

M = np.arange(16).reshape(4, 4)

print(M[[0, 2]])
print(M[[0, 2], [1, 3]])
M[[1, 3]] = -1

print(M)
```

## 6.  Aggregation along axes
Compute sum, mean, max along axis=0 and axis=1 of a 2D array. Use keepdims to preserve shape.

```python
M = np.array([[1, 2, 3], [4, 5, 6]])

print(M.sum(axis = 0))
print(M.sum(axis = 1))
print(M.sum())

print(M.sum(axis = 1))
print(M.sum(axis = 1, keepdims = True))

print(M.mean(axis =0))
print(M.max(axis = 1, keepdims = True))
```

## 7. Random number generation
Generate random integers, floats, and normally distributed data. Set a seed for reproducibility.

```python
import random

rng = np.random.default_rng(seed = 12)

print(rng.integers(0, 10, size = (3, 4)))

print(rng.random(size = (3, 3)))

print(rng.standard_normal(size = (3, 3)))

print(rng.normal(loc = 5.0, scale = 2.0, size = 100))

print(np.random.seed(42))
print(np.random.randint(0, 10, (3, 4)))
print(np.random.random((2, 5)))
print(np.random.randn(3, 3))
```

## 8. np.where & np.select
Use np.where to apply conditional logic. Use np.select for multiple condition branches.

```python
a = np.array([-3, 5, -1, 8, 0])

print(np.where(a > 0, a, 0))

print(np.where(a > 0, 'pos', 'non-pos'))

print(np.where(a > 0))

conditions = [
    a < 0,
    a == 0,
    a < 10
]

choices = ['negative', 'zero', 'positive']

print(np.select(conditions, choices, default = 'large'))
```

## 1. Matrix operations
Perform matrix multiplication using @, compute transpose, inverse, and determinant of a matrix.

```python
X = np.array(
    [[1, 2, 3],
    [4, 5, 8],
    [6, 7, 9]]
)

Y = np.array(
    [[3, 4, 5],
    [6, 9, 8],
    [1, 7, 2]]
)

print(X.T)

print("="*20)

print(Y.T)

print("="*20)

print(X @ Y)

print("="*20)

from numpy.linalg import inv, det

print(inv(X))

print("="*20)

print(inv(Y))

print("="*20)

print(det(X))

print("="*20)

print(det(Y))
```
